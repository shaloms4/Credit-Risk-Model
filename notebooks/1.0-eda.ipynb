1. Overview of the Data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display plots inline
%matplotlib inline

# Load your dataset (edit path as needed)
df = pd.read_csv("../data/raw/data.csv")
```

```python
# Shape of the dataset
print("Shape:", df.shape)

# Data types and non-null counts
df.info()

# First few rows
df.head()
```

2. Summary Statistics

```python
# Summary statistics for numerical columns
df.describe()
```

3. Distribution of Numerical Features

```python
numerical_cols = ['Amount', 'Value']

for col in numerical_cols:
    plt.figure(figsize=(8, 4))
    sns.histplot(df[col], kde=True, bins=50)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.grid(True)
    plt.show()
```

4. Distribution of Categorical Features

```python
categorical_cols = ['CurrencyCode', 'CountryCode', 'ProviderId', 'ProductCategory', 
                    'ChannelId', 'PricingStrategy', 'FraudResult']

for col in categorical_cols:
    plt.figure(figsize=(10, 4))
    sns.countplot(data=df, x=col, order=df[col].value_counts().index)
    plt.title(f'Count Plot of {col}')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

5. Correlation Analysis

```python
# Correlation matrix of numeric features
plt.figure(figsize=(8, 6))
corr_matrix = df.corr(numeric_only=True)
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title("Correlation Heatmap")
plt.show()
```

6. Identifying Missing Values

```python
# Count missing values
missing = df.isnull().sum()
missing = missing[missing > 0].sort_values(ascending=False)

# Display
print("Missing values:\n", missing)

# Only plot if there are missing values
if not missing.empty:
    plt.figure(figsize=(10, 6))
    missing.plot(kind='bar')
    plt.title("Missing Values per Feature")
    plt.ylabel("Count")
    plt.grid(True)
    plt.show()
else:
    print("No missing values found in the dataset.")
```

7. Outlier Detection with Box Plots

```python
for col in numerical_cols:
    plt.figure(figsize=(8, 4))
    sns.boxplot(x=df[col])
    plt.title(f'Boxplot of {col}')
    plt.grid(True)
    plt.show()
```
